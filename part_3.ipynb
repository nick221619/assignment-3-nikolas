```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
stereo_left = cv2.imread("assignment-3-data/stereo_left.jpg", cv2.IMREAD_GRAYSCALE)
plt.figure()
plt.imshow(stereo_left, cmap='gray')
print(stereo_left.shape, stereo_left.dtype)

stereo_right = cv2.imread("assignment-3-data/stereo_right.jpg", cv2.IMREAD_GRAYSCALE)
plt.figure()
plt.imshow(stereo_right, cmap='gray')
print(stereo_right.shape, stereo_right.dtype)
```

```python
def calculate_disparity_map(left, right, numDisparities, blockSize):
    stereo = cv2.StereoBM_create(numDisparities=numDisparities, blockSize=blockSize)
    disparity = stereo.compute(left, right)
    disparity = disparity.astype(np.float32) / 16.0
    return disparity
```

```python
disparity = calculate_disparity_map(stereo_left, stereo_right, 64, 15)

plt.figure(figsize=(10, 7))
plt.imshow(disparity, cmap='gray')
plt.colorbar(label="Disparity (pixels)")
plt.title("numDisparities=64, blockSize=15")
plt.savefig("assignment-3-data/outputs_3/disp_64_15.png")
plt.show()
```
